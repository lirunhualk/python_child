## 第 2 课：用变量保存信息

### 本节目标

- 知道变量可以保存信息。
- 会用变量保存名字、年龄和喜欢的东西。

```python
name = "猪妞"
age = 999999999999999999999999999999999999
favorite = "吃屎"

print("我的名字是", name)
print("我今年", age, "岁")
print("我喜欢", favorite)
```

### 小挑战

## 第 3 课：让电脑做判断

### 本节目标

- 知道 `if` 可以让电脑判断。
- 会根据分数输出不同结果。

```python
score = 85

if score >= 80:
    print("太棒了，继续保持！")
else:
    print("没关系，再练一次。")
```

### 试一试

把 `score` 改成不同的数字，观察电脑说的话有没有变化。

```python
score = 60

if score >= 90:
    print("超级厉害！")
elif score >= 70:
    print("已经不错了。")
else:
    print("再认真练一练。")
```

## 第 4 课：让电脑重复做事

### 本节目标

- 知道 `for` 可以重复执行代码。
- 会用循环打印多个数字或句子。

```python
for number in range(1, 6):
    print("第", number, "次：Python 真有趣！")
```

### 小挑战

把 `times` 改成 3、5、10，看看电脑会打印几次。

```python
times = 3

for i in range(times):
    print("我会写循环了！")
```

## 综合小项目：口算出题机

电脑可以帮我们自动出几道加法题。先运行看看，再修改题目数量和数字范围。

```python
import random

question_count = 3

for question_no in range(1, question_count + 1):
    a = random.randint(1, 20)
    b = random.randint(1, 20)
    print("第", question_no, "题：", a, "+", b, "=")
```
